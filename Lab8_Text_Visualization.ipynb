This notebook demonstrates advanced text visualization techniques including word clouds, sentiment-based visualizations, NER, POS tagging distribution, and sentiment polarity analysis using product reviews data.

**Key Libraries:** wordcloud, TextBlob, spaCy, matplotlib

## Setup

Install and import required packages, load spaCy model.

```python
# Install required packages (run once)
!pip install wordcloud matplotlib numpy pillow textblob spacy --quiet
!python -m spacy download en_core_web_sm --quiet

# Imports
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from wordcloud import WordCloud, STOPWORDS
from textblob import TextBlob
import spacy
from collections import Counter
import re
from spacy import displacy

# Load spaCy model
nlp = spacy.load("en_core_web_sm")
print('Setup complete!')
```

## Load Sample Product Reviews Dataset

Synthetic dataset of product reviews with sentiment labels (positive, negative, neutral).

```python
# Sample reviews dataset (text, sentiment_label)
reviews = [
    ("Excellent headphones with amazing sound quality and long battery life!", "positive"),
    ("Terrible laptop, crashes constantly and overheats.", "negative"),
    ("Phone works fine, battery life is average.", "neutral"),
    ("Love this tablet! Perfect for work and entertainment.", "positive"),
    ("Mouse stopped working after one week. Poor quality.", "negative"),
    ("Keyboard has standard features, nothing special.", "neutral"),
    ("Best monitor I've ever owned, crystal clear display.", "positive"),
    ("Camera takes blurry photos, very disappointed.", "negative"),
    # Add more reviews to make ~100 total (60% pos, 20% neg, 20% neut)
    ("Outstanding product, highly recommend!", "positive"),
    ("Complete waste of money, returned it.", "negative"),
    # ... (continuing pattern for demonstration - expand in full version)
] * 10  # Multiply to get sufficient data
np.random.shuffle(reviews)

print(f'Loaded {len(reviews)} reviews')
print('Sample:', reviews[:3])
```

## Task 1 – Word Cloud (Basic)

Generate word cloud from all review texts.

```python
# Join all review texts
all_text = ' '.join([review[0] for review in reviews])

# Generate word cloud
wordcloud = WordCloud(
    width=800, height=400, 
    background_color='white',
    stopwords=STOPWORDS
).generate(all_text)

# Display
plt.figure(figsize=(14, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.title('Word Cloud of Product Reviews', fontsize=16)
plt.axis('off')
plt.show()
```

## Task 2 – Sentiment-Based Word Clouds

Separate word clouds for positive, negative, and neutral reviews.

```python
# Separate reviews by sentiment
pos_reviews = ' '.join([r[0] for r in reviews if r[1] == 'positive'])
neg_reviews = ' '.join([r[0] for r in reviews if r[1] == 'negative'])
neu_reviews = ' '.join([r[0] for r in reviews if r[1] == 'neutral'])

# Generate word clouds
wc_pos = WordCloud(width=800, height=400, background_color='white', stopwords=STOPWORDS, colormap='Greens').generate(pos_reviews)
wc_neg = WordCloud(width=800, height=400, background_color='white', stopwords=STOPWORDS, colormap='Reds').generate(neg_reviews)
wc_neu = WordCloud(width=800, height=400, background_color='white', stopwords=STOPWORDS, colormap='Greys').generate(neu_reviews)

# Display side by side
fig, axes = plt.subplots(1, 3, figsize=(21, 6))
axes[0].imshow(wc_pos, interpolation='bilinear')
axes[0].set_title('Positive Reviews')
axes[0].axis('off')

axes[1].imshow(wc_neg, interpolation='bilinear')
axes[1].set_title('Negative Reviews')
axes[1].axis('off')

axes[2].imshow(wc_neu, interpolation='bilinear')
axes[2].set_title('Neutral Reviews')
axes[2].axis('off')

plt.suptitle('Word Clouds by Sentiment', fontsize=16)
plt.tight_layout()
plt.show()
```

## Task 3 – Named Entity Recognition (NER) with spaCy

Extract and visualize entity frequencies.

```python
# Process full text with spaCy
full_text = ' '.join([review[0] for review in reviews])
doc = nlp(full_text)

# Count entity labels
entity_counts = Counter([ent.label_ for ent in doc.ents])

# Plot horizontal bar chart
labels, counts = zip(*entity_counts.most_common())
plt.figure(figsize=(10, 6))
plt.barh(labels, counts, color='steelblue')
plt.xlabel('Count')
plt.ylabel('Entity Type')
plt.title('Named Entity Frequency in Reviews')
plt.tight_layout()
plt.show()

# Sample NER visualization
sample_doc = nlp(reviews[0][0])
displacy.render(sample_doc, style="ent", jupyter=True)
```

## Task 4 – POS Tag Distribution with spaCy

Distribution of major POS tags.

```python
# Extract POS tags from all reviews
pos_counter = Counter()
for review_text, _ in reviews:
    doc = nlp(review_text)
    pos_counter.update([token.pos_ for token in doc])

# Plot vertical bar chart
labels, counts = zip(*pos_counter.most_common())
colors = plt.cm.Set3(np.linspace(0, 1, len(labels)))
plt.figure(figsize=(12, 6))
bars = plt.bar(labels, counts, color=colors)
plt.xlabel('POS Tag')
plt.ylabel('Frequency')
plt.title('Part-of-Speech Distribution')

# Add value labels
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + max(counts)*0.01,
             f'{int(count)}', ha='center', va='bottom')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Task 5 – Sentiment Polarity Distribution

TextBlob polarity scores histogram.

```python
# Compute polarity scores
polarities = [TextBlob(review[0]).sentiment.polarity for review in reviews]

# Plot histogram
plt.figure(figsize=(10, 6))
plt.hist(polarities, bins=10, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(x=0, color='red', linestyle='--', linewidth=2, label='Neutral threshold')
plt.xlabel('Polarity Score')
plt.ylabel('Number of Reviews')
plt.title('Sentiment Polarity Distribution')
plt.legend()
plt.tight_layout()
plt.show()

print(f'Mean polarity: {np.mean(polarities):.3f}')
print(f'Std polarity: {np.std(polarities):.3f}')
```
