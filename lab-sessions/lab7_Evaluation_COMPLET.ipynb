```python
import numpy as np

df['Month'] = df['Date'].dt.month
month_names = ["Jan","Fév","Mar","Avr","Mai","Jun","Jul","Aoû","Sep","Oct","Nov","Déc"]

# Compute stats per month
stats = df.groupby('Month')['Temperature'].describe(percentiles=[.25,.5,.75])
stats.columns = ['count','mean','std','min','q25','q50','q75','max']
stats['iqr'] = stats['q75'] - stats['q25']
stats['upper'] = (stats['q75'] + 1.5 * stats['iqr']).clip(upper=stats['max'])
stats['lower'] = (stats['q25'] - 1.5 * stats['iqr']).clip(lower=stats['min'])
stats['month_name'] = month_names

src3 = ColumnDataSource(stats)
months_str = [str(m) for m in range(1,13)]

hover3 = HoverTool(tooltips=[
    ("Mois",   "@month_name"),
    ("Médiane","@q50{0.0} °C"),
    ("Min",    "@min{0.0} °C"),
    ("Max",    "@max{0.0} °C"),
])

p3 = figure(
    title="Distribution mensuelle des températures minimales",
    x_range=months_str, y_axis_label="Temperature (°C)",
    width=900, height=400,
    tools="pan,wheel_zoom,reset,save", toolbar_location="above"
)
p3.add_tools(hover3)

# Whiskers
p3.segment("month_name","upper","month_name","q75", source=src3, line_color="black")
p3.segment("month_name","lower","month_name","q25", source=src3, line_color="black")

# Boxes
p3.vbar("month_name", 0.7, "q75","q50", source=src3, fill_color="skyblue", line_color="black")
p3.vbar("month_name", 0.7, "q50","q25", source=src3, fill_color="lightcoral", line_color="black")

# Whisker caps
p3.rect("month_name","upper", 0.3, 0.01, source=src3, line_color="black")
p3.rect("month_name","lower", 0.3, 0.01, source=src3, line_color="black")

p3.xaxis.major_label_overrides = {str(i+1): month_names[i] for i in range(12)}
p3.xaxis.axis_label = "Mois"

show(p3)
```

## Question 5 : Sélection interactive de la plage temporelle

```python
from bokeh.models import DateRangeSlider, CustomJS

src5_full = ColumnDataSource(df)
src5_view = ColumnDataSource(df)

hover5 = HoverTool(tooltips=[
    ("Date","@Date{%F}"),
    ("Temp","@Temperature{0.0} °C"),
], formatters={"@Date":"datetime"}, mode="vline")

p5 = figure(
    title="Températures minimales — plage sélectionnable",
    x_axis_label="Date", y_axis_label="Temperature (°C)",
    x_axis_type="datetime", width=900, height=350,
    tools="pan,wheel_zoom,reset", toolbar_location="above"
)
p5.add_tools(hover5)
p5.line("Date","Temperature", source=src5_view, line_color="steelblue", line_width=1.2)
p5.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

slider = DateRangeSlider(
    title="Plage de dates",
    start=df['Date'].min(),
    end=df['Date'].max(),
    value=(df['Date'].min(), df['Date'].max()),
    step=1,
    width=900
)

callback = CustomJS(args=dict(src_full=src5_full, src_view=src5_view, slider=slider), code="""
    const [start, end] = slider.value;
    const full = src_full.data;
    const dates = full['Date'];
    const temps = full['Temperature'];

    const new_dates = [];
    const new_temps = [];
    for (let i = 0; i < dates.length; i++) {
        if (dates[i] >= start && dates[i] <= end) {
            new_dates.push(dates[i]);
            new_temps.push(temps[i]);
        }
    }
    src_view.data = {Date: new_dates, Temperature: new_temps};
    src_view.change.emit();
""")
slider.js_on_change('value', callback)

show(column(slider, p5))
```

```python
from bokeh.layouts import column as bk_column
# from bokeh.models import LinkedDataSource  ← ligne supprimée

# Rééchantillonage mensuel
df_monthly = df.set_index('Date').resample('MS')['Temperature'].mean().reset_index()
df_monthly.columns = ['Date','Monthly_Avg']

# Tendance : moyenne mobile 12 mois
df_monthly['Trend'] = df_monthly['Monthly_Avg'].rolling(window=12, center=True).mean()

# Saisonnalité
df_monthly['Seasonality'] = df_monthly['Monthly_Avg'] - df_monthly['Trend']

src6 = ColumnDataSource(df_monthly)

kwargs = dict(x_axis_type="datetime", width=900, height=220,
              tools="pan,wheel_zoom,reset", toolbar_location="above")

# Plot 1 : données mensuelles
p6a = figure(title="Données mensuelles", x_axis_label="", y_axis_label="Temp (°C)", **kwargs)
p6a.line("Date","Monthly_Avg", source=src6, color="steelblue", line_width=2)

# Plot 2 : tendance
p6b = figure(title="Tendance (moy. mobile 12 mois)", x_range=p6a.x_range,
             x_axis_label="", y_axis_label="Temp (°C)", **kwargs)
p6b.line("Date","Trend", source=src6, color="firebrick", line_width=2)

# Plot 3 : saisonnalité
p6c = figure(title="Saisonnalité", x_range=p6a.x_range,
             x_axis_label="Date", y_axis_label="Delta (°C)", **kwargs)
p6c.line("Date","Seasonality", source=src6, color="seagreen", line_width=2)
p6c.line([df_monthly["Date"].min(), df_monthly["Date"].max()],[0,0],
         color="black", line_dash="dashed", line_width=1)

for p in [p6a, p6b, p6c]:
    p.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(bk_column(p6a, p6b, p6c))
```
