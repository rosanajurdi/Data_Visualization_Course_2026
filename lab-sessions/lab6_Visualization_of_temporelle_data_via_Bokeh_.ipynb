**Introduction to Bokeh**
Bokeh is an interactive visualization library for Python that targets modern web browsers for presentation.
Unlike Matplotlib, which is primarily designed for static plots, Bokeh excels at creating
interactive plots and dashboards. It can handle large datasets and streaming data,
making it suitable for real-time applications.

**Key Features of Bokeh:**

* **Interactivity:** Built-in support for zooming, panning, hovering, and other interactive tools.
* **Web-Focused:** Generates HTML and JavaScript, making it easy to embed plots in web pages.
* **High Performance:** Can handle large datasets efficiently.
* **Versatility:** Supports a wide range of plot types (lines, bars, scatter plots, etc.).

## 3. Time Series Line Plot

Basic interactive line plot of monthly passenger counts over time.

```python
source = ColumnDataSource(data={
    "date":       df.index,
    "passengers": df["Passengers"],
})

p = figure(
    title="Air Passengers – Monthly Count (1949-1960)",
    x_axis_type="datetime",
    width=900, height=400,
    tools="pan,wheel_zoom,box_zoom,reset,save",
    toolbar_location="above",
)

p.line("date", "passengers", source=source,
       line_width=2, color="steelblue", legend_label="Passengers")
p.circle("date", "passengers", source=source,
         size=5, color="steelblue", alpha=0.5)

p.add_tools(HoverTool(
    tooltips=[
        ("Date",       "@date{%b %Y}"),
        ("Passengers", "@passengers{0,0}"),
    ],
    formatters={"@date": "datetime"},
    mode="vline"
))

p.xaxis.axis_label = "Date"
p.yaxis.axis_label = "Number of Passengers"
p.xaxis.formatter = DatetimeTickFormatter(months="%b %Y", years="%Y")
p.legend.location = "top_left"
p.grid.grid_line_alpha = 0.3

show(p)
```

## 4. Bar Plot – Average Passengers by Month

Aggregated bar chart showing the average number of passengers per calendar month.

```python
month_names = ["Jan","Feb","Mar","Apr","May","Jun",
               "Jul","Aug","Sep","Oct","Nov","Dec"]

monthly_avg = df.groupby(df.index.month)["Passengers"].mean()
monthly_std = df.groupby(df.index.month)["Passengers"].std()

src_bar = ColumnDataSource(data={
    "months":  month_names,
    "avg":     monthly_avg.values,
    "std":     monthly_std.values,
    "upper":   (monthly_avg + monthly_std).values,
    "lower":   (monthly_avg - monthly_std).values,
})

p2 = figure(
    title="Average Passengers by Month",
    x_range=month_names,
    width=800, height=400,
    tools="pan,wheel_zoom,box_zoom,reset,save",
    toolbar_location="above",
)

p2.vbar(x="months", top="avg", source=src_bar,
        width=0.7,
        color=factor_cmap("months", palette=Spectral6 + Spectral6, factors=month_names),
        line_color="white", alpha=0.85, legend_label="Mean")

# Error bars (std)
p2.segment(x0="months", y0="lower", x1="months", y1="upper",
           source=src_bar, line_width=2, color="black")
p2.rect(x="months", y="upper", width=0.2, height=0.5,
        source=src_bar, color="black")
p2.rect(x="months", y="lower", width=0.2, height=0.5,
        source=src_bar, color="black")

p2.add_tools(HoverTool(tooltips=[
    ("Month",  "@months"),
    ("Mean",   "@avg{0.0}"),
    ("± Std",  "@std{0.0}"),
]))

p2.xaxis.axis_label = "Month"
p2.yaxis.axis_label = "Average Passengers"
p2.legend.location = "top_left"
p2.grid.grid_line_alpha = 0.3

show(p2)
```

## 5. Mean & Standard Deviation Band

Rolling mean ± 1 standard deviation displayed as a shaded band.

```python
window = 12   # 12-month rolling window

roll_mean = df["Passengers"].rolling(window=window).mean()
roll_std  = df["Passengers"].rolling(window=window).std()

src_band = ColumnDataSource(data={
    "date":    df.index,
    "raw":     df["Passengers"],
    "mean":    roll_mean,
    "upper":   roll_mean + roll_std,
    "lower":   roll_mean - roll_std,
})

p3 = figure(
    title=f"Rolling Mean ± 1 Std (window={window} months)",
    x_axis_type="datetime",
    width=900, height=420,
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

# Shaded band
band = Band(base="date", lower="lower", upper="upper",
            source=src_band, fill_color="steelblue", fill_alpha=0.25,
            line_color="steelblue", line_alpha=0.4)
p3.add_layout(band)

# Raw series
p3.line("date", "raw", source=src_band,
        line_width=1.5, color="lightsteelblue", alpha=0.7, legend_label="Raw")
# Rolling mean
p3.line("date", "mean", source=src_band,
        line_width=2.5, color="darkblue", legend_label="Rolling Mean")

p3.add_tools(HoverTool(
    tooltips=[
        ("Date",         "@date{%b %Y}"),
        ("Raw",          "@raw{0,0}"),
        ("Rolling Mean", "@mean{0.0}"),
        ("Upper Band",   "@upper{0.0}"),
        ("Lower Band",   "@lower{0.0}"),
    ],
    formatters={"@date": "datetime"},
    mode="vline"
))

p3.xaxis.axis_label  = "Date"
p3.yaxis.axis_label  = "Passengers"
p3.xaxis.formatter   = DatetimeTickFormatter(months="%b %Y", years="%Y")
p3.legend.location   = "top_left"
p3.grid.grid_line_alpha = 0.3

show(p3)
```

## 6. Outlier Detection

Points that fall outside **mean ± 2 standard deviations** (global) are flagged as outliers and highlighted in red.

```python
global_mean = df["Passengers"].mean()
global_std  = df["Passengers"].std()
threshold   = 2

upper_bound = global_mean + threshold * global_std
lower_bound = global_mean - threshold * global_std

is_outlier  = (df["Passengers"] > upper_bound) | (df["Passengers"] < lower_bound)
normal_df   = df[~is_outlier]
outlier_df  = df[is_outlier]

src_normal  = ColumnDataSource(data={"date": normal_df.index, "pass": normal_df["Passengers"]})
src_outlier = ColumnDataSource(data={"date": outlier_df.index, "pass": outlier_df["Passengers"]})

p4 = figure(
    title="Outlier Detection (Mean ± 2 Std)",
    x_axis_type="datetime",
    width=900, height=420,
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

# Shaded outlier zones
p4.add_layout(BoxAnnotation(bottom=upper_bound, fill_alpha=0.08, fill_color="red"))
p4.add_layout(BoxAnnotation(top=lower_bound,    fill_alpha=0.08, fill_color="red"))

# Threshold lines
p4.add_layout(Span(location=upper_bound, dimension="width",
                   line_color="red", line_dash="dashed", line_width=1.5))
p4.add_layout(Span(location=lower_bound, dimension="width",
                   line_color="red", line_dash="dashed", line_width=1.5))
p4.add_layout(Span(location=global_mean,  dimension="width",
                   line_color="green", line_dash="dotted", line_width=1.5))

# Normal points
p4.line("date", "pass", source=src_normal,
        line_width=2, color="steelblue", legend_label="Normal")
p4.circle("date", "pass", source=src_normal,
          size=5, color="steelblue", alpha=0.6)

# Outlier points
p4.circle("date", "pass", source=src_outlier,
          size=10, color="red", alpha=0.9, legend_label="Outlier")

p4.add_tools(HoverTool(
    tooltips=[("Date", "@date{%b %Y}"), ("Passengers", "@pass{0,0}")],
    formatters={"@date": "datetime"}
))

p4.xaxis.axis_label = "Date"
p4.yaxis.axis_label = "Passengers"
p4.xaxis.formatter  = DatetimeTickFormatter(months="%b %Y", years="%Y")
p4.legend.location  = "top_left"
p4.grid.grid_line_alpha = 0.3

print(f"Global mean:   {global_mean:.1f}")
print(f"Upper bound:   {upper_bound:.1f}")
print(f"Lower bound:   {lower_bound:.1f}")
print(f"Outliers found: {is_outlier.sum()}")
if not outlier_df.empty:
    display(outlier_df)

show(p4)
```

## 7. Resampling

### 7a. Downsampling – Annual Mean

Resample the monthly data to a yearly frequency (mean).

```python
# --- Downsampling: yearly mean ---
df_yearly = df.resample("YE").mean()

src_yearly = ColumnDataSource(data={
    "date": df_yearly.index,
    "pass": df_yearly["Passengers"],
})

p5 = figure(
    title="Downsampling – Yearly Mean Passengers",
    x_axis_type="datetime",
    width=700, height=380,
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

p5.vbar(x="date", top="pass", source=src_yearly,
        width=2.5e7, color="coral", alpha=0.8, legend_label="Yearly Mean")
p5.line("date", "pass", source=src_yearly,
        line_width=2, color="darkred", legend_label="Trend")

p5.add_tools(HoverTool(
    tooltips=[("Year", "@date{%Y}"), ("Mean Passengers", "@pass{0.0}")],
    formatters={"@date": "datetime"}
))

p5.xaxis.axis_label = "Year"
p5.yaxis.axis_label = "Mean Passengers"
p5.xaxis.formatter  = DatetimeTickFormatter(years="%Y")
p5.legend.location  = "top_left"
p5.grid.grid_line_alpha = 0.3

show(p5)
```

### 7b. Upsampling – Weekly with Linear Interpolation

Upsample from monthly to **weekly** frequency using linear interpolation.

```python
# --- Upsampling: weekly with linear interpolation ---
df_weekly = df.resample("W").interpolate(method="linear")

src_monthly = ColumnDataSource(data={"date": df.index,        "pass": df["Passengers"]})
src_weekly  = ColumnDataSource(data={"date": df_weekly.index, "pass": df_weekly["Passengers"]})

p6 = figure(
    title="Upsampling – Weekly (Linear Interpolation) vs Monthly",
    x_axis_type="datetime",
    width=900, height=420,
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

p6.line("date", "pass", source=src_weekly,
        line_width=1.5, color="orange", alpha=0.7, legend_label="Weekly (interpolated)")
p6.circle("date", "pass", source=src_monthly,
          size=8, color="navy", alpha=0.9, legend_label="Monthly (original)")

p6.add_tools(HoverTool(
    tooltips=[("Date", "@date{%d %b %Y}"), ("Passengers", "@pass{0.0}")],
    formatters={"@date": "datetime"}
))

p6.xaxis.axis_label = "Date"
p6.yaxis.axis_label = "Passengers"
p6.xaxis.formatter  = DatetimeTickFormatter(months="%b %Y", years="%Y")
p6.legend.location  = "top_left"
p6.grid.grid_line_alpha = 0.3

print(f"Monthly points: {len(df)}   →   Weekly points: {len(df_weekly)}")
show(p6)
```

## 8. Lag Analysis

Scatter plot of the series against itself shifted by **k** lags (months).  
A strong linear pattern indicates high autocorrelation at that lag.

```python
lags_to_plot = [1, 6, 12, 24]

lag_figures = []
for lag in lags_to_plot:
    lagged   = df["Passengers"].shift(lag).dropna()
    original = df["Passengers"].loc[lagged.index]

    src_lag = ColumnDataSource(data={
        "x":    original.values,
        "y":    lagged.values,
        "date": original.index,
    })

    # Correlation coefficient for title
    corr = np.corrcoef(original.values, lagged.values)[0, 1]

    plag = figure(
        title=f"Lag {lag} month(s)  –  r = {corr:.3f}",
        width=350, height=350,
        tools="pan,wheel_zoom,reset",
    )
    plag.circle("x", "y", source=src_lag,
                size=6, color="steelblue", alpha=0.6)

    # Reference line y = x
    mn, mx = original.min(), original.max()
    plag.line([mn, mx], [mn, mx], line_dash="dashed", color="red", line_width=1.5)

    plag.add_tools(HoverTool(
        tooltips=[
            ("Date",         "@date{%b %Y}"),
            ("t",            "@x{0,0}"),
            (f"t-{lag}",     "@y{0,0}"),
        ],
        formatters={"@date": "datetime"}
    ))

    plag.xaxis.axis_label = "Passengers at t"
    plag.yaxis.axis_label = f"Passengers at t-{lag}"
    plag.grid.grid_line_alpha = 0.3
    lag_figures.append(plag)

grid = gridplot([lag_figures[:2], lag_figures[2:]])
show(grid)
```

## 9. Autocorrelation (ACF)

The **Autocorrelation Function (ACF)** measures the correlation between the series and its own lagged values.  
Significant spikes every **12 lags** reveal the yearly seasonality of the data.

```python
from statsmodels.tsa.stattools import acf

n_lags = 40
acf_values, conf_int = acf(df["Passengers"], nlags=n_lags, alpha=0.05)

lags   = np.arange(n_lags + 1)
ci_up  = conf_int[:, 1] - acf_values
ci_low = acf_values - conf_int[:, 0]

# Significance threshold ≈ 1.96 / sqrt(n)
sig_threshold = 1.96 / np.sqrt(len(df))

colors = ["crimson" if abs(v) > sig_threshold else "steelblue" for v in acf_values]

src_acf = ColumnDataSource(data={
    "lag":    lags,
    "acf":    acf_values,
    "color":  colors,
    "ci_up":  ci_up,
    "ci_low": ci_low,
})

p7 = figure(
    title="Autocorrelation Function (ACF)",
    width=900, height=420,
    x_range=Range1d(-0.5, n_lags + 0.5),
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

# Confidence interval band
p7.add_layout(BoxAnnotation(
    bottom=-sig_threshold, top=sig_threshold,
    fill_color="lightblue", fill_alpha=0.2,
    line_color="steelblue", line_dash="dashed"
))

# Vertical stems
p7.segment(x0="lag", y0=0, x1="lag", y1="acf",
           source=src_acf, line_width=2, line_color="color")
p7.circle("lag", "acf", source=src_acf,
          size=8, color="color", alpha=0.9)

# Zero baseline
p7.add_layout(Span(location=0, dimension="width",
                   line_color="black", line_width=1))

p7.add_tools(HoverTool(tooltips=[
    ("Lag",  "@lag"),
    ("ACF",  "@acf{0.000}"),
]))

p7.xaxis.axis_label = "Lag (months)"
p7.yaxis.axis_label = "Autocorrelation"
p7.grid.grid_line_alpha = 0.3

print(f"Significance threshold: ±{sig_threshold:.4f}")
show(p7)
```

## 10. Yearly Trend Comparison

Each year's monthly profile plotted on the same axis to visualise the growth in passenger numbers over the decade.

```python
years  = sorted(df.index.year.unique())
palette = Blues8[::-1]  # light → dark

p8 = figure(
    title="Monthly Profile per Year",
    x_range=month_names,
    width=900, height=420,
    tools="pan,wheel_zoom,box_zoom,reset,save",
)

for i, year in enumerate(years):
    year_df = df[df.index.year == year].copy()
    year_df["month_name"] = year_df.index.month.map(
        lambda m: month_names[m - 1]
    )
    year_df = year_df.set_index("month_name").reindex(month_names)

    src_y = ColumnDataSource(data={
        "month":  month_names,
        "pass":   year_df["Passengers"].values,
    })

    color = palette[i % len(palette)]
    p8.line("month", "pass", source=src_y,
            line_width=2, color=color, alpha=0.85, legend_label=str(year))
    p8.circle("month", "pass", source=src_y,
              size=5, color=color, alpha=0.7)

p8.add_tools(HoverTool(tooltips=[
    ("Month",      "@month"),
    ("Passengers", "@pass{0,0}"),
]))

p8.xaxis.axis_label = "Month"
p8.yaxis.axis_label = "Passengers"
p8.legend.location  = "top_left"
p8.legend.label_text_font_size = "9pt"
p8.grid.grid_line_alpha = 0.3

show(p8)
```

## 11. Summary Dashboard

Side-by-side overview: raw series + rolling mean (top) and yearly bar chart (bottom).
