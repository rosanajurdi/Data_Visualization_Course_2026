# Lab 4: MPG Dataset Analysis with Seaborn

This lab explores the MPG dataset using various Seaborn visualization techniques including scatter plots, line plots, and heatmaps to understand relationships between vehicle features and fuel efficiency.

## Part 1 – Scatter Plots

Scatter plots are ideal for visualizing relationships between two continuous variables. We'll create several scatter plots to explore how MPG relates to different vehicle characteristics.

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
mpg_df = sns.load_dataset("mpg")

# Drop rows with NaN values
mpg_df = mpg_df.dropna()

# Print basic information
print(mpg_df.head())
print(mpg_df.info())
print(mpg_df.describe())
```

### Plot 1 – Basic Scatter: MPG vs Weight

This scatter plot shows the relationship between vehicle weight and miles per gallon. We expect heavier vehicles to have lower fuel efficiency.

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=mpg_df, x='weight', y='mpg')
plt.title("MPG vs Weight")
plt.xlabel("Weight")
plt.ylabel("Miles Per Gallon")
plt.tight_layout()
plt.show()
```

### Plot 2 – Scatter with Hue: MPG vs Horsepower by Origin

Adding hue by origin allows us to compare fuel efficiency vs horsepower across different manufacturing regions (usa, europe, japan).

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=mpg_df, x='horsepower', y='mpg', hue='origin')
plt.title("MPG vs Horsepower by Origin")
plt.xlabel("Horsepower")
plt.ylabel("Miles Per Gallon")
plt.legend(title='Origin')
plt.tight_layout()
plt.show()
```

### Plot 3 – Scatter with Size and Hue: MPG vs Displacement by Cylinders

Using both hue and size for cylinders creates a rich visualization showing how engine size affects displacement and MPG.

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=mpg_df, x='displacement', y='mpg', hue='cylinders', size='cylinders', palette='viridis')
plt.title("MPG vs Displacement by Number of Cylinders")
plt.xlabel("Displacement")
plt.ylabel("Miles Per Gallon")
plt.tight_layout()
plt.show()
```

### Plot 4 – Regression Scatter (lmplot): MPG vs Weight by Origin

lmplot adds regression lines to show the trend and confidence intervals for each origin group.

```python
sns.lmplot(data=mpg_df, x='weight', y='mpg', hue='origin', height=5, aspect=1.5)
plt.suptitle("MPG vs Weight with Regression Lines by Origin")
plt.tight_layout()
plt.show()
```

## Part 2 – Line Plots

Line plots are perfect for showing trends over time or ordered categories. Here we examine how MPG and other features evolved over model years.

### Plot 5 – Average MPG Over Model Years

This shows the overall trend in fuel efficiency improvement over time.

```python
plt.figure(figsize=(10, 5))
mpg_yearly = mpg_df.groupby('model_year')['mpg'].mean().reset_index()
plt.plot(mpg_yearly['model_year'], mpg_yearly['mpg'], marker='o')
plt.title("Average MPG Over Model Years")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Plot 6 – Average MPG Over Model Years by Origin

Comparing MPG trends across different regions shows regional differences in fuel efficiency progress.

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=mpg_df, x='model_year', y='mpg', hue='origin', marker=True, dashes=False)
plt.title("Average MPG Over Model Years by Origin")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.legend(title='Origin')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Plot 7 – Multiple Continuous Features Over Model Years

Normalized features allow comparison of different scales on the same plot to see parallel trends.

```python
# Group by model_year and compute means
features = ['mpg', 'horsepower', 'weight', 'acceleration']
yearly_stats = mpg_df.groupby('model_year')[features].mean().reset_index()

# Min-max normalization
for feature in features:
    min_val = yearly_stats[feature].min()
    max_val = yearly_stats[feature].max()
    yearly_stats[f'{feature}_norm'] = (yearly_stats[feature] - min_val) / (max_val - min_val)

# Plot
plt.figure(figsize=(12, 6))
plt.plot(yearly_stats['model_year'], yearly_stats['mpg_norm'], marker='o', label='MPG')
plt.plot(yearly_stats['model_year'], yearly_stats['horsepower_norm'], marker='s', label='Horsepower')
plt.plot(yearly_stats['model_year'], yearly_stats['weight_norm'], marker='^', label='Weight')
plt.plot(yearly_stats['model_year'], yearly_stats['acceleration_norm'], marker='d', label='Acceleration')

plt.title("Normalized Feature Trends Over Model Years")
plt.xlabel("Model Year")
plt.ylabel("Normalized Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Part 3 – Heatmaps

Heatmaps excel at showing correlations and patterns in pivot tables with color intensity.

### Plot 8 – Correlation Heatmap

The correlation matrix reveals relationships between all continuous features.

```python
numeric_cols = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration']
corr_matrix = mpg_df[numeric_cols].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix of MPG Dataset Features")
plt.tight_layout()
plt.show()
```

### Plot 9 – Pivot Heatmap: Average MPG by Cylinders and Origin

This heatmap reveals which origin produces the most efficient engines for each cylinder count.

```python
mpg_pivot1 = mpg_df.pivot_table(index='cylinders', columns='origin', values='mpg', aggfunc='mean')

plt.figure(figsize=(8, 5))
sns.heatmap(mpg_pivot1, annot=True, fmt='.1f', cmap='YlGnBu')
plt.title("Average MPG by Number of Cylinders and Origin")
plt.xlabel("Origin")
plt.ylabel("Cylinders")
plt.tight_layout()
plt.show()
```

### Plot 10 – Pivot Heatmap: Average Horsepower by Model Year and Origin

This shows horsepower evolution patterns by origin across model years.

```python
hp_pivot = mpg_df.pivot_table(index='model_year', columns='origin', values='horsepower', aggfunc='mean')

plt.figure(figsize=(8, 10))
sns.heatmap(hp_pivot, annot=True, fmt='.1f', cmap='OrRd')
plt.title("Average Horsepower by Model Year and Origin")
plt.xlabel("Origin")
plt.ylabel("Model Year")
plt.tight_layout()
plt.show()
```
