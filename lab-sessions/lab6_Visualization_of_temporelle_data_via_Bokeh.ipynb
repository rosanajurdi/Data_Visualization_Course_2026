**Introduction to Bokeh**
Bokeh is an interactive visualization library for Python that targets modern web browsers for presentation.
Unlike Matplotlib, which is primarily designed for static plots, Bokeh excels at creating
interactive plots and dashboards. It can handle large datasets and streaming data,
making it suitable for real-time applications.

**Key Features of Bokeh:**

* **Interactivity:** Built-in support for zooming, panning, hovering, and other interactive tools.
* **Web-Focused:** Generates HTML and JavaScript, making it easy to embed plots in web pages.
* **High Performance:** Can handle large datasets efficiently.
* **Versatility:** Supports a wide range of plot types (lines, bars, scatter plots, etc.).

## Section 2: Date/Time Manipulation
<hr style="border: 2px solid black;">

We convert the `Date` column to proper `datetime` objects so Bokeh can render them on
a time axis. We also extract useful date components (month, day, day name) for later
aggregations.
<hr style="border: 2px solid black;">

```python
# %%
# Section 2: Date/Time Manipulation
# ---------------------------------

# 1. Convert to Datetime
# ~~~~~~~~~~~~~~~~~~~~~

passenger_df["Date"] = pd.to_datetime(passenger_df["Date"])

# Verify the conversion
print(passenger_df.info())

# 2. Extract Date Components
# ~~~~~~~~~~~~~~~~~~~~~~~~~

passenger_df["Month"]    = passenger_df["Date"].dt.month
passenger_df["Day"]      = passenger_df["Date"].dt.day
passenger_df["Day_Name"] = passenger_df["Date"].dt.day_name()

print(passenger_df.head())
```

## Section 3: Time Series Visualization and Analysis
<hr style="border: 2px solid black;">

### 3.1 Basic Time Series Line Plot
We use `figure()` with a `datetime` x-axis type and add a `line` glyph.
A `HoverTool` is added to display the exact date and passenger count on hover.
<hr style="border: 2px solid black;">

```python
# %%
# Section 3: Time Series Visualization and Analysis
# -------------------------------------------------

# 1. Basic Time Series Plot
# ~~~~~~~~~~~~~~~~~~~~~~~~

# Bokeh works best with a ColumnDataSource for data binding
source = ColumnDataSource(passenger_df)

# Create the figure with a datetime x-axis
p1 = figure(
    title="Air Passengers Over Time",
    x_axis_label="Date",
    y_axis_label="Number of Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Add the line glyph
p1.line(
    x="Date", y="#Passengers",
    source=source,
    line_width=2,
    color="steelblue",
    legend_label="Passengers"
)

# Add circle markers at each data point
p1.circle(
    x="Date", y="#Passengers",
    source=source,
    size=4,
    color="steelblue",
    alpha=0.6
)

# Add a HoverTool for interactivity
hover1 = HoverTool(
    tooltips=[
        ("Date",       "@Date{%F}"),
        ("Passengers", "@{#Passengers}")
    ],
    formatters={"@Date": "datetime"},
    mode="vline"
)
p1.add_tools(hover1)

# Styling
p1.legend.location = "top_left"
p1.grid.grid_line_alpha = 0.4
p1.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(p1)
```

### 3.2 Aggregation and Bar Plot

We group passengers by month and create an interactive vertical bar chart using
Bokeh's `vbar` glyph. Month numbers are treated as categorical values on the x-axis.
<hr style="border: 2px solid black;">

```python
# 2. Aggregation and Bar Plot
# ~~~~~~~~~~~~~~~~~~~~~~~~~

passengers_per_month = (
    passenger_df.groupby("Month")["#Passengers"].sum().reset_index()
)

# Convert month numbers to strings for categorical axis
month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
passengers_per_month["Month_Name"] = passengers_per_month["Month"].apply(
    lambda m: month_names[m - 1]
)

source_bar = ColumnDataSource(passengers_per_month)

p2 = figure(
    title="Total Passengers per Month",
    x_axis_label="Month",
    y_axis_label="Total Passengers",
    x_range=month_names,          # Categorical x-axis in calendar order
    width=850, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

p2.vbar(
    x="Month_Name",
    top="#Passengers",
    source=source_bar,
    width=0.7,
    color=factor_cmap(
        "Month_Name", palette=Category20[12], factors=month_names
    ),
    line_color="white",
    legend_field="Month_Name"
)

# HoverTool
hover2 = HoverTool(
    tooltips=[
        ("Month",          "@Month_Name"),
        ("Total Passengers", "@{#Passengers}{0,0}")
    ]
)
p2.add_tools(hover2)

p2.legend.location = "top_left"
p2.legend.label_text_font_size = "9pt"
p2.grid.grid_line_alpha = 0.4
p2.xgrid.grid_line_color = None

show(p2)
```

### 3.3 Mean and Standard Deviation

We overlay horizontal reference lines for the mean and ±1 standard deviation using
Bokeh's `Span` model and `add_layout`. This makes it easy to judge how far each
observation deviates from the average.
<hr style="border: 2px solid black;">

```python
# 3. Mean and Standard Deviation
# ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~

mean_passengers = passenger_df["#Passengers"].mean()
std_passengers  = passenger_df["#Passengers"].std()

source3 = ColumnDataSource(passenger_df)

p3 = figure(
    title="Passengers with Mean and Standard Deviation",
    x_axis_label="Date",
    y_axis_label="Number of Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Main line
line_main = p3.line(
    x="Date", y="#Passengers",
    source=source3,
    line_width=2, color="steelblue",
    legend_label="Passengers"
)

# Span for mean
mean_span = Span(
    location=mean_passengers,
    dimension="width",
    line_color="red",
    line_dash="dashed",
    line_width=2
)

# Spans for mean ± std
upper_span = Span(
    location=mean_passengers + std_passengers,
    dimension="width",
    line_color="green",
    line_dash="dashed",
    line_width=2
)
lower_span = Span(
    location=mean_passengers - std_passengers,
    dimension="width",
    line_color="green",
    line_dash="dashed",
    line_width=2
)

p3.add_layout(mean_span)
p3.add_layout(upper_span)
p3.add_layout(lower_span)

# Manual legend entries for the spans
from bokeh.models import Line

p3.line([], [], line_color="red",   line_dash="dashed", line_width=2,
        legend_label=f"Mean: {mean_passengers:.2f}")
p3.line([], [], line_color="green", line_dash="dashed", line_width=2,
        legend_label=f"Mean ± Std ({mean_passengers - std_passengers:.2f} / {mean_passengers + std_passengers:.2f})")

# HoverTool
hover3 = HoverTool(
    tooltips=[
        ("Date",       "@Date{%F}"),
        ("Passengers", "@{#Passengers}")
    ],
    formatters={"@Date": "datetime"},
    mode="vline"
)
p3.add_tools(hover3)

p3.legend.location = "top_left"
p3.grid.grid_line_alpha = 0.4
p3.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(p3)
```

## Section 4: Outlier Detection
<hr style="border: 2px solid black;">

### 4.1 Z-Score Calculation
We compute a Z-score for every observation. Points whose absolute Z-score exceeds 2
are considered outliers and highlighted in red in the next plot.
<hr style="border: 2px solid black;">

```python
# %%
# Section 4: Outlier Detection
# ----------------------------

# 1. Z-Score Calculation
# ~~~~~~~~~~~~~~~~~~~~

passenger_df["Z-Score"] = (
    passenger_df["#Passengers"] - passenger_df["#Passengers"].mean()
) / passenger_df["#Passengers"].std()

passenger_df["Absolute_Z-Score"] = abs(passenger_df["Z-Score"])

print(passenger_df.sort_values(by="Absolute_Z-Score", ascending=False).head(10))
```

### 4.2 Outlier Visualization

We draw the full time series as a line and overlay red circles on the outlier points.
The `HoverTool` shows the Z-score alongside the date and passenger count.
<hr style="border: 2px solid black;">

```python
# 2. Outlier Visualization
# ~~~~~~~~~~~~~~~~~~~~~~~

outliers     = passenger_df[passenger_df["Absolute_Z-Score"] > 2]
non_outliers = passenger_df[passenger_df["Absolute_Z-Score"] <= 2]

source_full     = ColumnDataSource(passenger_df)
source_outliers = ColumnDataSource(outliers)

p4 = figure(
    title="Air Passengers with Outliers (|Z-Score| > 2)",
    x_axis_label="Date",
    y_axis_label="Number of Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Full time series
p4.line(
    x="Date", y="#Passengers",
    source=source_full,
    line_width=2, color="steelblue",
    legend_label="Passengers"
)

# Outlier circles
p4.circle(
    x="Date", y="#Passengers",
    source=source_outliers,
    size=10, color="red", alpha=0.8,
    legend_label="Outliers (|Z| > 2)"
)

# HoverTool – activates on the circles
hover4 = HoverTool(
    tooltips=[
        ("Date",        "@Date{%F}"),
        ("Passengers",  "@{#Passengers}"),
        ("Z-Score",     "@{Z-Score}{0.00}")
    ],
    formatters={"@Date": "datetime"}
)
p4.add_tools(hover4)

p4.legend.location = "top_left"
p4.grid.grid_line_alpha = 0.4
p4.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(p4)
```

## Section 5: Resampling
<hr style="border: 2px solid black;">

### 5.1 Upsampling (monthly → daily with linear interpolation)

We set `Date` as the DataFrame index and use `resample('D').asfreq()` followed by
`interpolate(method='linear')` to fill daily gaps. Both the upsampled and original
series are overlaid on the same Bokeh figure.
<hr style="border: 2px solid black;">

```python
# %%
# Section 5: Resampling
# ----------------------

# Set 'Date' as index for resampling operations
passenger_df.set_index("Date", inplace=True)

# 1. Upsampling
# ~~~~~~~~~~~~

# Upsample to daily frequency
daily_passengers = passenger_df.resample("D").asfreq()

# Interpolate missing values linearly
daily_passengers["#Passengers"] = daily_passengers["#Passengers"].interpolate(method="linear")

# Prepare sources
source_daily    = ColumnDataSource(dict(
    date=daily_passengers.index,
    passengers=daily_passengers["#Passengers"]
))
source_original = ColumnDataSource(dict(
    date=passenger_df.index,
    passengers=passenger_df["#Passengers"]
))

p5 = figure(
    title="Upsampling to Daily Frequency (with Linear Interpolation)",
    x_axis_label="Date",
    y_axis_label="Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Upsampled daily line
p5.line(
    x="date", y="passengers",
    source=source_daily,
    line_width=1.5, color="orange", line_dash="dashed",
    legend_label="Upsampled & Interpolated (Daily)"
)

# Original monthly line
p5.line(
    x="date", y="passengers",
    source=source_original,
    line_width=2.5, color="steelblue", alpha=0.8,
    legend_label="Original (Monthly)"
)
p5.circle(
    x="date", y="passengers",
    source=source_original,
    size=5, color="steelblue", alpha=0.9
)

hover5 = HoverTool(
    tooltips=[
        ("Date",       "@date{%F}"),
        ("Passengers", "@passengers{0.0}")
    ],
    formatters={"@date": "datetime"},
    mode="vline"
)
p5.add_tools(hover5)

p5.legend.location = "top_left"
p5.grid.grid_line_alpha = 0.4
p5.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(p5)
```

### 5.2 Downsampling (monthly → yearly average)

We use `resample('YE')` (Year-End) and compute the mean for each year, then overlay
it with the original monthly series. Circle markers on the yearly average help
spot the trend at a glance.
<hr style="border: 2px solid black;">

```python
# 2. Downsampling
# ~~~~~~~~~~~~~~

# Downsample to yearly average
yearly_passengers = passenger_df.resample("YE")["#Passengers"].mean()

source_yearly = ColumnDataSource(dict(
    date=yearly_passengers.index,
    passengers=yearly_passengers.values
))

p6 = figure(
    title="Downsampling to Yearly Frequency (Mean)",
    x_axis_label="Date",
    y_axis_label="Average Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Original monthly
p6.line(
    x="date", y="passengers",
    source=source_original,
    line_width=1.5, color="steelblue", alpha=0.5,
    legend_label="Original (Monthly)"
)

# Yearly average
p6.line(
    x="date", y="passengers",
    source=source_yearly,
    line_width=2.5, color="crimson",
    legend_label="Yearly Average"
)
p6.circle(
    x="date", y="passengers",
    source=source_yearly,
    size=8, color="crimson"
)

hover6 = HoverTool(
    tooltips=[
        ("Year",       "@date{%Y}"),
        ("Avg Passengers", "@passengers{0.0}")
    ],
    formatters={"@date": "datetime"}
)
p6.add_tools(hover6)

p6.legend.location = "top_left"
p6.grid.grid_line_alpha = 0.4
p6.xaxis.formatter = DatetimeTickFormatter(years="%Y")

show(p6)
```

## Section 6: Shift and Lag Analysis
<hr style="border: 2px solid black;">

We use `shift()` to create a one-period lag copy of the passenger series (`#Passengers_Shift`)
and a frequency-based shift (`#Passengers_tShift`). Overlaying original and shifted series
highlights how much the time series auto-correlates with itself at lag 1.
<hr style="border: 2px solid black;">

```python
# %%
# Section 6: Shift and tshift (Lag Analysis)
# -----------------------------------------

# Reset index to access 'Date' as a column again
passenger_df.reset_index(inplace=True)

# Create shifted series (value shift by 1 position)
passenger_df["#Passengers_Shift"] = passenger_df["#Passengers"].shift(periods=1)

# Frequency-based tshift: requires DatetimeIndex
# Set Date as index temporarily, shift the index by 1 month, then extract values
df_indexed = passenger_df.set_index("Date")
df_tshifted = df_indexed[["#Passengers"]].shift(periods=1, freq="MS")
# Merge back on Date
df_tshifted = df_tshifted.reset_index().rename(columns={"#Passengers": "#Passengers_tShift"})
passenger_df = passenger_df.merge(df_tshifted, on="Date", how="left")

print(passenger_df[["Date", "#Passengers", "#Passengers_Shift", "#Passengers_tShift"]].head(8))

# --- Bokeh visualization ---

source_shift = ColumnDataSource(passenger_df.dropna(subset=["#Passengers_Shift"]))

p7 = figure(
    title="Lag Analysis: Original vs Shifted (shift=1) vs tShifted (freq=MS)",
    x_axis_label="Date",
    y_axis_label="Number of Passengers",
    x_axis_type="datetime",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Original
p7.line(
    x="Date", y="#Passengers",
    source=source_shift,
    line_width=2, color="steelblue",
    legend_label="Original"
)

# Value shift (lag-1: values shift down, index unchanged)
p7.line(
    x="Date", y="#Passengers_Shift",
    source=source_shift,
    line_width=2, color="orange", line_dash="dashed",
    legend_label="Shift (lag-1 value)"
)

# Frequency-based tshift
p7.line(
    x="Date", y="#Passengers_tShift",
    source=source_shift,
    line_width=2, color="green", line_dash="dotdash",
    legend_label="tShift (index +1 month)"
)

hover7 = HoverTool(
    tooltips=[
        ("Date",     "@Date{%F}"),
        ("Original", "@{#Passengers}"),
        ("Shift",    "@{#Passengers_Shift}{0.0}"),
        ("tShift",   "@{#Passengers_tShift}{0.0}")
    ],
    formatters={"@Date": "datetime"},
    mode="vline"
)
p7.add_tools(hover7)

p7.legend.location = "top_left"
p7.grid.grid_line_alpha = 0.4
p7.xaxis.formatter = DatetimeTickFormatter(months="%b %Y")

show(p7)
```

## Section 7: Autocorrelation (ACF)
<hr style="border: 2px solid black;">

Bokeh does not have a built-in ACF plot like Matplotlib/statsmodels, so we:
1. Compute ACF values manually with `statsmodels.tsa.stattools.acf`.
2. Compute the 95 % confidence band: `±1.96 / √N`.
3. Draw vertical bars (one per lag) with segments from zero to the ACF value,
   and a shaded confidence band using `varea`.

The result is a fully interactive ACF chart equivalent to `plot_acf()`.
<hr style="border: 2px solid black;">

```python
# %%
# Section 7: Autocorrelation
# --------------------------

from statsmodels.tsa.stattools import acf

# Re-use the original passenger series (ensure no NaNs)
series = passenger_df["#Passengers"].dropna()

n_lags = 30
acf_values, confint = acf(series, nlags=n_lags, alpha=0.05)

lags       = np.arange(0, n_lags + 1)
conf_band  = 1.96 / np.sqrt(len(series))   # symmetric 95% confidence
upper_conf = confint[:, 1] - acf_values    # upper margin
lower_conf = confint[:, 0] - acf_values    # lower margin (negative)

# Determine bar colour: significant lags in blue, others in grey
colors = ["steelblue" if abs(v) > conf_band else "lightgrey"
          for v in acf_values]

source_acf = ColumnDataSource(dict(
    lags=lags,
    acf=acf_values,
    upper=acf_values + upper_conf,
    lower=acf_values + lower_conf,
    colors=colors
))

p8 = figure(
    title="Autocorrelation Function (ACF) — up to lag 30",
    x_axis_label="Lag",
    y_axis_label="Autocorrelation",
    width=900, height=450,
    tools="pan,wheel_zoom,box_zoom,reset,save"
)

# Shaded 95 % confidence band
p8.varea(
    x=lags,
    y1= conf_band,
    y2=-conf_band,
    color="lightblue", alpha=0.4,
    legend_label="95% Confidence Band"
)

# Zero baseline
p8.line(
    x=[lags[0], lags[-1]], y=[0, 0],
    line_color="black", line_width=1
)

# Vertical segments from 0 to ACF value
p8.segment(
    x0="lags", y0=0,
    x1="lags", y1="acf",
    source=source_acf,
    line_width=3,
    line_color="colors"
)

# Circles at the tip of each bar
p8.circle(
    x="lags", y="acf",
    source=source_acf,
    size=8,
    color="colors"
)

# HoverTool
hover8 = HoverTool(
    tooltips=[
        ("Lag",           "@lags"),
        ("ACF",           "@acf{0.000}"),
        ("95% CI upper",  "@upper{0.000}"),
        ("95% CI lower",  "@lower{0.000}")
    ]
)
p8.add_tools(hover8)

p8.legend.location = "top_right"
p8.grid.grid_line_alpha = 0.4
p8.y_range.start = -1.1
p8.y_range.end   =  1.1

show(p8)
```
